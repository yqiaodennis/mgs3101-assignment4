## Step 1: Set up and create the data

Run both cells. The fixed random seed makes the data reproducible. Each row is one week.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.model_selection import train_test_split

plt.style.use('seaborn-v0_8-whitegrid')
rng = np.random.default_rng(42)
print('Setup complete.')
````

````python
advertising_spend = np.linspace(1, 10, 80)
noise = rng.normal(loc=0, scale=7, size=80)
weekly_sales = 18 + 5.5 * advertising_spend + noise

data = pd.DataFrame({
    'advertising_spend_thousands': advertising_spend,
    'weekly_sales_thousands': weekly_sales,
})

print(f'Number of observations: {len(data)}')
data.head()
````

**Answer 1:** Identify the response variable, the predictor, the units, and what one row represents.

`Response variable`: Weekly sales revenue.

`Predictor`: Weekly advertising expenditure.

`Units`: Both variables are measured in thousands of dollars.

`Each row`: represents one week's data.

The training set is used to enable the model to learn the relationship between advertising expenditure and sales revenue, while the testing set is used to test the model's predictive ability on data that did not participate in the training. If a model is trained using a test set, it cannot objectively evaluate its predictive ability for new data, which may lead to overly optimistic evaluation results.

## Step 4: Fit the baseline and regression models

- **Baseline:** predicts the training-set average sales for every test week.
- **Linear regression:** uses advertising spend to predict sales.

````python
baseline_model = DummyRegressor(strategy='mean')
baseline_model.fit(X_train, y_train)
baseline_predictions = baseline_model.predict(X_test)

regression_model = LinearRegression()
regression_model.fit(X_train, y_train)
regression_predictions = regression_model.predict(X_test)

print(f'Baseline predicts the training mean: {y_train.mean():.2f} ($ thousands)')
print(f'Regression intercept: {regression_model.intercept_:.2f} ($ thousands)')
print(f'Regression slope: {regression_model.coef_[0]:.2f} ($ thousands of sales per $1,000 spend)')
````

````python
# For test size = 0.20, random_state = 17
baseline_model = DummyRegressor(strategy='mean')
baseline_model.fit(X_train, y_train)
baseline_predictions = baseline_model.predict(X_test)

regression_model = LinearRegression()
regression_model.fit(X_train, y_train)
regression_predictions = regression_model.predict(X_test)

print(f'Baseline predicts the training mean: {y_train.mean():.2f} ($ thousands)')
print(f'Regression intercept: {regression_model.intercept_:.2f} ($ thousands)')
print(f'Regression slope: {regression_model.coef_[0]:.2f} ($ thousands of sales per $1,000 spend)')
````

````python
# For test size = 0.25, random_state = 7
baseline_model = DummyRegressor(strategy='mean')
baseline_model.fit(X_train, y_train)
baseline_predictions = baseline_model.predict(X_test)

regression_model = LinearRegression()
regression_model.fit(X_train, y_train)
regression_predictions = regression_model.predict(X_test)

print(f'Baseline predicts the training mean: {y_train.mean():.2f} ($ thousands)')
print(f'Regression intercept: {regression_model.intercept_:.2f} ($ thousands)')
print(f'Regression slope: {regression_model.coef_[0]:.2f} ($ thousands of sales per $1,000 spend)')
````

**Answer 4:** Interpret the slope with correct units. Explain what the baseline predicts and why it is a useful comparison.

`Baseline model` always predict the average sales value of the training set, which is $47.63 thousand. It does not consider changes in advertising spending and serves as a simple reference standard to determine whether the linear regression model has truly improved its predictive ability.

`Slope` is 5.49, meaning that for every $1000 increase in advertising spending, the model predicts an average increase of $5490 in weekly sales. This indicates a positive linear relationship between the two, but cannot prove that advertising expenditure leads to an increase in sales.

`Intercept` is 18.03, meaning that when advertising expenditure is zero, the model predicts a weekly sales revenue of $18.03 thousand.

The prediction of baseline model is not the sales corresponding to a specific advertising expenditure level, but the average weekly sales of the training set, which is $47.63 thousand. Therefore, regardless of the advertising expenditure for a particular week, Baseline predicts a sales revenue of $47.63 thousand.

It is a simple reference model that does not require consideration of the relationship between advertising expenditure and sales revenue. Comparing the linear regression model with the baseline can determine whether the variable of advertising expenditure truly provides additional value for sales forecasting. If the linear regression model has a lower prediction error on the test set, it indicates that it has more predictive value than simply predicting average sales.

## Step 5: Compare test performance

MAE and RMSE are in $ thousands of sales. Lower is better. RMSE gives more weight to large errors.

````python
results = pd.DataFrame({
    'model': ['Training-mean baseline', 'Linear regression'],
    'test_MAE_thousands': [
        mean_absolute_error(y_test, baseline_predictions),
        mean_absolute_error(y_test, regression_predictions),
    ],
    'test_RMSE_thousands': [
        np.sqrt(mean_squared_error(y_test, baseline_predictions)),
        np.sqrt(mean_squared_error(y_test, regression_predictions)),
    ],
})

results.round(2)
````

````python
# For test size = 0.20, random_state = 17
results = pd.DataFrame({
    'model': ['Training-mean baseline', 'Linear regression'],
    'test_MAE_thousands': [
        mean_absolute_error(y_test, baseline_predictions),
        mean_absolute_error(y_test, regression_predictions),
    ],
    'test_RMSE_thousands': [
        np.sqrt(mean_squared_error(y_test, baseline_predictions)),
        np.sqrt(mean_squared_error(y_test, regression_predictions)),
    ],
})

results.round(2)
````

````python
# For test size = 0.25, random_state = 7
results = pd.DataFrame({
    'model': ['Training-mean baseline', 'Linear regression'],
    'test_MAE_thousands': [
        mean_absolute_error(y_test, baseline_predictions),
        mean_absolute_error(y_test, regression_predictions),
    ],
    'test_RMSE_thousands': [
        np.sqrt(mean_squared_error(y_test, baseline_predictions)),
        np.sqrt(mean_squared_error(y_test, regression_predictions)),
    ],
})

results.round(2)
````

````python
comparison = pd.DataFrame({
    'actual_sales_thousands': y_test.to_numpy(),
    'baseline_prediction_thousands': baseline_predictions,
    'regression_prediction_thousands': regression_predictions,
}).sort_values('actual_sales_thousands').reset_index(drop=True)

plt.figure(figsize=(9, 5))
plt.plot(comparison.index, comparison['actual_sales_thousands'], 'o', label='Actual test sales')
plt.plot(comparison.index, comparison['baseline_prediction_thousands'], 'x', label='Baseline prediction')
plt.plot(comparison.index, comparison['regression_prediction_thousands'], '.', label='Regression prediction')
plt.xlabel('Test observations (sorted by actual sales)')
plt.ylabel('Sales ($ thousands)')
plt.title('Actual and predicted sales on the test set')
plt.legend()
plt.show()
````
